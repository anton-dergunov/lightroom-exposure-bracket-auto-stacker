```python
import exiftool
import glob
import pandas as pd
from IPython.display import display
```

```python
ATTRIBUTES = [
    "SourceFile",
    "EXIF:Make",
    "EXIF:DateTimeOriginal",
    "EXIF:ExposureMode",
    "EXIF:ExposureCompensation",
    "MakerNotes:ReleaseMode",
    "MakerNotes:SequenceImageNumber",
    "MakerNotes:SequenceLength",
]
```

```python
def show_exif(path_spec):
    files = glob.glob(path_spec)

    with exiftool.ExifToolHelper() as et:
        metadata = et.get_metadata(files)

    filtered_metadata = [{key: item.get(key, None) for key in ATTRIBUTES} for item in metadata]
    df = pd.DataFrame(filtered_metadata)

    df.sort_values(by="SourceFile", inplace=True)
    df.reset_index(drop=True, inplace=True)

    with pd.option_context('display.max_columns', None, 'display.max_rows', None, 'display.max_colwidth', None):
        display(df)
```
